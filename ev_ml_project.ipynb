```python
# ================================================================
# ADVANCED EV SALES FORECASTING USING MACHINE LEARNING
# ================================================================
# Project Features:
# 1. Data Cleaning
# 2. Exploratory Data Analysis (EDA)
# 3. Feature Engineering
# 4. PCA
# 5. Five Machine Learning Models
# 6. Time-Based Train/Test Split
# 7. Cross-Validation
# 8. Hyperparameter Tuning
# 9. Model Evaluation
# 10. Feature Importance
# 11. Model Saving
# 12. New EV Sales Prediction
# ================================================================


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.base import clone

from sklearn.compose import ColumnTransformer

from sklearn.decomposition import PCA

from sklearn.impute import SimpleImputer

from sklearn.pipeline import Pipeline

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.model_selection import (
    TimeSeriesSplit,
    RandomizedSearchCV
)

from sklearn.linear_model import (
    Ridge,
    ElasticNet
)

from sklearn.ensemble import (
    RandomForestRegressor,
    ExtraTreesRegressor,
    HistGradientBoostingRegressor
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ================================================================
# 2. LOAD DATASET
# ================================================================

FILE_PATH = "/content/EV_Dataset.csv"

df = pd.read_csv(FILE_PATH)

print("Original Dataset Shape:", df.shape)

display(df.head())

print("\nDataset Information:")
df.info()


# ================================================================
# 3. DATA CLEANING
# ================================================================

df = df.drop_duplicates().copy()

# Convert Date column
df["Date"] = pd.to_datetime(
    df["Date"],
    errors="coerce"
)

# Convert numeric columns
df["Year"] = pd.to_numeric(
    df["Year"],
    errors="coerce"
)

df["EV_Sales_Quantity"] = pd.to_numeric(
    df["EV_Sales_Quantity"],
    errors="coerce"
)

# Required columns
required_columns = [
    "Date",
    "Year",
    "Month_Name",
    "State",
    "Vehicle_Class",
    "Vehicle_Category",
    "Vehicle_Type",
    "EV_Sales_Quantity"
]

# Remove missing values
df = df.dropna(
    subset=required_columns
).copy()

# Remove negative sales
df = df[
    df["EV_Sales_Quantity"] >= 0
].copy()

# Clean categorical columns
categorical_columns = [
    "Month_Name",
    "State",
    "Vehicle_Class",
    "Vehicle_Category",
    "Vehicle_Type"
]

for column in categorical_columns:

    df[column] = (
        df[column]
        .astype(str)
        .str.strip()
        .str.lower()
    )

# Sort data chronologically
df = df.sort_values(
    "Date"
).reset_index(drop=True)

print("\nCleaned Dataset Shape:", df.shape)

display(df.head())


# ================================================================
# 4. FEATURE ENGINEERING
# ================================================================

# Date-based features

df["Date_Year"] = df["Date"].dt.year

df["Date_Month"] = df["Date"].dt.month

df["Date_Quarter"] = df["Date"].dt.quarter

df["Date_Day"] = df["Date"].dt.day

df["Date_DayOfWeek"] = df["Date"].dt.dayofweek


# Cyclic month features

df["Month_Sin"] = np.sin(
    2 * np.pi * df["Date_Month"] / 12
)

df["Month_Cos"] = np.cos(
    2 * np.pi * df["Date_Month"] / 12
)


# Cyclic quarter features

df["Quarter_Sin"] = np.sin(
    2 * np.pi * df["Date_Quarter"] / 4
)

df["Quarter_Cos"] = np.cos(
    2 * np.pi * df["Date_Quarter"] / 4
)


# Log transformation

df["Target_Log"] = np.log1p(
    df["EV_Sales_Quantity"]
)


print("\nFeature Engineering Completed")

display(df.head())


# ================================================================
# 5. EXPLORATORY DATA ANALYSIS
# ================================================================

print("\nDescriptive Statistics:")

display(
    df.describe(include="all")
)


# ------------------------------------------------
# 5.1 Distribution of EV Sales
# ------------------------------------------------

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="EV_Sales_Quantity",
    bins=50,
    kde=True
)

plt.title(
    "Distribution of EV Sales Quantity"
)

plt.xlabel(
    "EV Sales Quantity"
)

plt.ylabel(
    "Frequency"
)

plt.show()


# ------------------------------------------------
# 5.2 Total Sales by Year
# ------------------------------------------------

year_summary = (
    df.groupby(
        "Year",
        as_index=False
    )["EV_Sales_Quantity"]
    .sum()
)

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=year_summary,
    x="Year",
    y="EV_Sales_Quantity",
    marker="o"
)

plt.title(
    "Total EV Sales by Year"
)

plt.xlabel("Year")

plt.ylabel("Total EV Sales")

plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------
# 5.3 Top 10 States by Sales
# ------------------------------------------------

state_summary = (
    df.groupby(
        "State",
        as_index=False
    )["EV_Sales_Quantity"]
    .sum()
    .sort_values(
        "EV_Sales_Quantity",
        ascending=False
    )
    .head(10)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=state_summary,
    x="EV_Sales_Quantity",
    y="State"
)

plt.title(
    "Top 10 States by EV Sales"
)

plt.xlabel("Total EV Sales")

plt.ylabel("State")

plt.show()


# ------------------------------------------------
# 5.4 Sales by Vehicle Category
# ------------------------------------------------

category_summary = (
    df.groupby(
        "Vehicle_Category",
        as_index=False
    )["EV_Sales_Quantity"]
    .sum()
    .sort_values(
        "EV_Sales_Quantity",
        ascending=False
    )
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=category_summary,
    x="Vehicle_Category",
    y="EV_Sales_Quantity"
)

plt.title(
    "EV Sales by Vehicle Category"
)

plt.xlabel("Vehicle Category")

plt.ylabel("Total EV Sales")

plt.xticks(rotation=45)

plt.show()


# ================================================================
# 6. SELECT FEATURES AND TARGET
# ================================================================

TARGET = "EV_Sales_Quantity"

FEATURES = [
    "Year",
    "Month_Name",
    "Date_Year",
    "Date_Month",
    "Date_Quarter",
    "Date_Day",
    "Date_DayOfWeek",
    "Month_Sin",
    "Month_Cos",
    "Quarter_Sin",
    "Quarter_Cos",
    "State",
    "Vehicle_Class",
    "Vehicle_Category",
    "Vehicle_Type"
]

# Keep only available columns

FEATURES = [
    column
    for column in FEATURES
    if column in df.columns
]

X = df[FEATURES].copy()

y = df[TARGET].copy()

print("\nSelected Features:")

print(FEATURES)

print("\nTarget Column:")

print(TARGET)


# ================================================================
# 7. TIME-BASED TRAIN-TEST SPLIT
# ================================================================

# Since this is time-related data,
# we avoid random shuffling.

split_index = int(
    len(df) * 0.80
)

X_train = X.iloc[
    :split_index
].copy()

X_test = X.iloc[
    split_index:
].copy()

y_train = y.iloc[
    :split_index
].copy()

y_test = y.iloc[
    split_index:
].copy()

print("\nTraining Rows:", len(X_train))

print("Testing Rows:", len(X_test))

print(
    "\nTraining Date Range:",
    df["Date"].iloc[0],
    "to",
    df["Date"].iloc[split_index - 1]
)

print(
    "\nTesting Date Range:",
    df["Date"].iloc[split_index],
    "to",
    df["Date"].iloc[-1]
)


# ================================================================
# 8. ADVANCED PREPROCESSING WITH PCA
# ================================================================

# Identify numeric features

numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()


# Identify categorical features

categorical_features = X_train.select_dtypes(
    include=["object"]
).columns.tolist()


print("\nNumeric Features:")

print(numeric_features)

print("\nCategorical Features:")

print(categorical_features)


# ------------------------------------------------
# Numeric Pipeline
# ------------------------------------------------

# Steps:
# 1. Fill missing values
# 2. Standardize values
# 3. Apply PCA

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        ),

        (
            "pca",
            PCA(
                n_components=0.95,
                random_state=42
            )
        )
    ]
)


# ------------------------------------------------
# Categorical Pipeline
# ------------------------------------------------

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)


# ------------------------------------------------
# Combined Preprocessor
# ------------------------------------------------

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric_pca",
            numeric_pipeline,
            numeric_features
        ),

        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ],

    remainder="drop"
)


print("\nPCA Preprocessing Pipeline Created")


# ================================================================
# 9. DEFINE FIVE MACHINE LEARNING MODELS
# ================================================================

models = {

    "Ridge Regression": Ridge(
        alpha=1.0
    ),

    "Elastic Net": ElasticNet(
        alpha=0.001,
        l1_ratio=0.5,
        max_iter=3000,
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=150,
        max_depth=25,
        min_samples_split=2,
        random_state=42,
        n_jobs=-1
    ),

    "Extra Trees": ExtraTreesRegressor(
        n_estimators=150,
        max_depth=25,
        min_samples_split=2,
        random_state=42,
        n_jobs=-1
    ),

    "Histogram Gradient Boosting":
        HistGradientBoostingRegressor(
            max_iter=200,
            learning_rate=0.08,
            max_leaf_nodes=31,
            l2_regularization=0.1,
            random_state=42
        )
}


print("\nFive Models Defined:")

for model_name in models:
    print("-", model_name)


# ================================================================
# 10. TRAIN AND EVALUATE FIVE MODELS
# ================================================================

results = []

trained_models = {}

predictions = {}


for model_name, model in models.items():

    print("\n" + "=" * 75)

    print(
        "TRAINING MODEL:",
        model_name
    )

    print("=" * 75)


    # Create pipeline

    pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),

            (
                "model",
                model
            )
        ]
    )


    # Train model

    pipeline.fit(
        X_train,
        y_train
    )


    # Make predictions

    y_pred = pipeline.predict(
        X_test
    )


    # Avoid negative sales predictions

    y_pred = np.maximum(
        y_pred,
        0
    )


    # Evaluation metrics

    mae = mean_absolute_error(
        y_test,
        y_pred
    )


    rmse = np.sqrt(
        mean_squared_error(
            y_test,
            y_pred
        )
    )


    r2 = r2_score(
        y_test,
        y_pred
    )


    # Save results

    results.append(
        {
            "Model": model_name,
            "MAE": mae,
            "RMSE": rmse,
            "R2 Score": r2
        }
    )


    # Save trained model

    trained_models[
        model_name
    ] = pipeline


    # Save predictions

    predictions[
        model_name
    ] = y_pred


    print(
        "MAE:",
        round(mae, 4)
    )

    print(
        "RMSE:",
        round(rmse, 4)
    )

    print(
        "R2 Score:",
        round(r2, 4)
    )


# ================================================================
# 11. MODEL COMPARISON
# ================================================================

results_df = pd.DataFrame(
    results
)


# Sort by RMSE
# Lower RMSE is better

results_df = results_df.sort_values(
    by="RMSE",
    ascending=True
).reset_index(
    drop=True
)


print("\nFINAL MODEL COMPARISON:")

display(
    results_df
)


# ------------------------------------------------
# RMSE Comparison
# ------------------------------------------------

plt.figure(figsize=(12, 6))

sns.barplot(
    data=results_df,
    x="RMSE",
    y="Model"
)

plt.title(
    "Machine Learning Model Comparison - RMSE"
)

plt.xlabel("RMSE")

plt.ylabel("Model")

plt.show()


# ------------------------------------------------
# R2 Comparison
# ------------------------------------------------

plt.figure(figsize=(12, 6))

sns.barplot(
    data=results_df,
    x="R2 Score",
    y="Model"
)

plt.title(
    "Machine Learning Model Comparison - R2 Score"
)

plt.xlabel("R2 Score")

plt.ylabel("Model")

plt.show()


# ------------------------------------------------
# MAE Comparison
# ------------------------------------------------

plt.figure(figsize=(12, 6))

sns.barplot(
    data=results_df,
    x="MAE",
    y="Model"
)

plt.title(
    "Machine Learning Model Comparison - MAE"
)

plt.xlabel("MAE")

plt.ylabel("Model")

plt.show()


# ================================================================
# 12. SELECT BEST MODEL
# ================================================================

# Best model based on lowest RMSE

best_model_name = results_df.iloc[
    0
]["Model"]

best_model = trained_models[
    best_model_name
]

best_predictions = predictions[
    best_model_name
]


print(
    "\nBest Model Based on Lowest RMSE:"
)

print(
    best_model_name
)


# ================================================================
# 13. TIME SERIES CROSS-VALIDATION
# ================================================================

# TimeSeriesSplit preserves chronological order.

tscv = TimeSeriesSplit(
    n_splits=3
)

cv_rmse_scores = []


for fold, (
    train_index,
    validation_index
) in enumerate(
    tscv.split(X_train),
    start=1
):

    print(
        "\nCross-Validation Fold:",
        fold
    )


    X_fold_train = X_train.iloc[
        train_index
    ]

    X_fold_validation = X_train.iloc[
        validation_index
    ]


    y_fold_train = y_train.iloc[
        train_index
    ]

    y_fold_validation = y_train.iloc[
        validation_index
    ]


    # Clone selected model

    fold_model = clone(
        best_model
    )


    # Train fold model

    fold_model.fit(
        X_fold_train,
        y_fold_train
    )


    # Predict validation data

    fold_prediction = fold_model.predict(
        X_fold_validation
    )


    fold_prediction = np.maximum(
        fold_prediction,
        0
    )


    # Calculate RMSE

    fold_rmse = np.sqrt(
        mean_squared_error(
            y_fold_validation,
            fold_prediction
        )
    )


    cv_rmse_scores.append(
        fold_rmse
    )


    print(
        "Fold RMSE:",
        round(fold_rmse, 4)
    )


print(
    "\nAverage Cross-Validation RMSE:",
    round(
        np.mean(
            cv_rmse_scores
        ),
        4
    )
)


# ================================================================
# 14. HYPERPARAMETER TUNING
# ================================================================

# Tune Extra Trees using RandomizedSearchCV

tuning_pipeline = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),

        (
            "model",
            ExtraTreesRegressor(
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)


# Hyperparameter search space

parameter_grid = {

    "model__n_estimators": [
        100,
        150,
        200
    ],

    "model__max_depth": [
        None,
        15,
        25,
        35
    ],

    "model__min_samples_split": [
        2,
        5,
        10
    ],

    "model__min_samples_leaf": [
        1,
        2,
        4
    ],

    "model__max_features": [
        1.0,
        "sqrt",
        0.7
    ]
}


# Randomized search

random_search = RandomizedSearchCV(

    estimator=tuning_pipeline,

    param_distributions=parameter_grid,

    n_iter=8,

    scoring="neg_root_mean_squared_error",

    cv=3,

    verbose=1,

    random_state=42,

    n_jobs=-1
)


print(
    "\nStarting Hyperparameter Tuning..."
)


random_search.fit(
    X_train,
    y_train
)


# Get best tuned model

tuned_model = random_search.best_estimator_


# Make tuned predictions

tuned_prediction = tuned_model.predict(
    X_test
)


tuned_prediction = np.maximum(
    tuned_prediction,
    0
)


# Evaluate tuned model

tuned_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        tuned_prediction
    )
)


tuned_mae = mean_absolute_error(
    y_test,
    tuned_prediction
)


tuned_r2 = r2_score(
    y_test,
    tuned_prediction
)


print(
    "\nBEST HYPERPARAMETERS:"
)

print(
    random_search.best_params_
)


print(
    "\nTUNED EXTRA TREES RESULTS:"
)

print(
    "MAE:",
    round(tuned_mae, 4)
)

print(
    "RMSE:",
    round(tuned_rmse, 4)
)

print(
    "R2 Score:",
    round(tuned_r2, 4)
)


# ================================================================
# 15. SELECT FINAL MODEL
# ================================================================

# Use tuned model if it performs better.

if tuned_rmse < results_df.iloc[0]["RMSE"]:

    final_model = tuned_model

    final_model_name = "Tuned Extra Trees"

    final_prediction = tuned_prediction

else:

    final_model = best_model

    final_model_name = best_model_name

    final_prediction = best_predictions


print(
    "\nFINAL SELECTED MODEL:"
)

print(
    final_model_name
)


# ================================================================
# 16. ACTUAL VS PREDICTED VALUES
# ================================================================

comparison_df = pd.DataFrame(

    {
        "Actual Sales": y_test.values,

        "Predicted Sales": final_prediction
    }

)


comparison_df["Residual"] = (

    comparison_df["Actual Sales"]

    - comparison_df["Predicted Sales"]

)


print(
    "\nActual vs Predicted Values:"
)

display(
    comparison_df.head(20)
)


# Plot actual vs predicted

plt.figure(figsize=(8, 8))

plt.scatter(

    comparison_df["Actual Sales"],

    comparison_df["Predicted Sales"],

    alpha=0.35

)


minimum_value = min(

    comparison_df["Actual Sales"].min(),

    comparison_df["Predicted Sales"].min()

)


maximum_value = max(

    comparison_df["Actual Sales"].max(),

    comparison_df["Predicted Sales"].max()

)


plt.plot(

    [
        minimum_value,
        maximum_value
    ],

    [
        minimum_value,
        maximum_value
    ],

    linestyle="--"

)


plt.title(
    "Actual vs Predicted EV Sales"
)

plt.xlabel(
    "Actual Sales"
)

plt.ylabel(
    "Predicted Sales"
)

plt.show()


# ================================================================
# 17. RESIDUAL ANALYSIS
# ================================================================

plt.figure(figsize=(12, 5))

sns.histplot(

    comparison_df["Residual"],

    bins=50,

    kde=True

)


plt.title(
    "Residual Distribution"
)

plt.xlabel(
    "Residual"
)

plt.ylabel(
    "Frequency"
)

plt.show()


# ================================================================
# 18. FEATURE IMPORTANCE
# ================================================================

# Feature importance is available for tree-based models.

if final_model_name == "Tuned Extra Trees":

    importance_pipeline = tuned_model

else:

    importance_pipeline = trained_models[
        "Extra Trees"
    ]


tree_model = importance_pipeline.named_steps[
    "model"
]


fitted_preprocessor = (
    importance_pipeline.named_steps[
        "preprocessor"
    ]
)


feature_names = (
    fitted_preprocessor
    .get_feature_names_out()
)


feature_importances = (
    tree_model.feature_importances_
)


importance_df = pd.DataFrame(

    {
        "Feature": feature_names,

        "Importance": feature_importances
    }

).sort_values(

    by="Importance",

    ascending=False

).head(25)


print(
    "\nTOP 25 IMPORTANT FEATURES:"
)

display(
    importance_df
)


plt.figure(figsize=(12, 9))

sns.barplot(

    data=importance_df,

    x="Importance",

    y="Feature"

)


plt.title(
    "Top 25 Feature Importances"
)

plt.xlabel(
    "Importance"
)

plt.ylabel(
    "Feature"
)

plt.show()


# ================================================================
# 19. SAVE FINAL MODEL
# ================================================================

MODEL_FILE = (
    "final_ev_sales_advanced_model.pkl"
)

RESULTS_FILE = (
    "ev_advanced_model_comparison.csv"
)


# Save model and metadata

joblib.dump(

    {

        "model_name": final_model_name,

        "model_pipeline": final_model,

        "features": FEATURES,

        "target": TARGET

    },

    MODEL_FILE

)


# Save model comparison

results_df.to_csv(

    RESULTS_FILE,

    index=False

)


print(
    "\nSaved Model:"
)

print(
    MODEL_FILE
)


print(
    "\nSaved Results:"
)

print(
    RESULTS_FILE
)


# ================================================================
# 20. NEW EV SALES PREDICTION
# ================================================================

# Change these values for your prediction.

new_data = pd.DataFrame(

    {

        "Year": [2024],

        "Month_Name": ["jan"],

        "Date_Year": [2024],

        "Date_Month": [1],

        "Date_Quarter": [1],

        "Date_Day": [1],

        "Date_DayOfWeek": [0],

        "Month_Sin": [
            np.sin(
                2 * np.pi * 1 / 12
            )
        ],

        "Month_Cos": [
            np.cos(
                2 * np.pi * 1 / 12
            )
        ],

        "Quarter_Sin": [
            np.sin(
                2 * np.pi * 1 / 4
            )
        ],

        "Quarter_Cos": [
            np.cos(
                2 * np.pi * 1 / 4
            )
        ],

        "State": ["karnataka"],

        "Vehicle_Class": ["bus"],

        "Vehicle_Category": ["bus"],

        "Vehicle_Type": ["bus"]

    }

)


# Keep same feature order

new_data = new_data[
    FEATURES
]


# Predict new sales

new_prediction = final_model.predict(
    new_data
)


new_prediction = max(

    float(
        new_prediction[0]
    ),

    0

)


print(
    "\nPredicted EV Sales Quantity:"
)

print(
    round(
        new_prediction,
        2
    )
)


# ================================================================
# 21. FINAL PROJECT SUMMARY
# ================================================================

print(
    "\n" + "=" * 75
)

print(
    "ADVANCED EV SALES PROJECT COMPLETED"
)

print(
    "=" * 75
)

print(
    "Dataset Rows:",
    len(df)
)

print(
    "Number of Models:",
    5
)

print(
    "PCA Included: Yes"
)

print(
    "Time-Based Split: Yes"
)

print(
    "Cross-Validation: Yes"
)

print(
    "Hyperparameter Tuning: Yes"
)

print(
    "Final Model:",
    final_model_name
)

print(
    "Target:",
    TARGET
)

print(
    "=" * 75
)
```
